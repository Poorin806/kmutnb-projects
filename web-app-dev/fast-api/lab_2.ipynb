```python
from fastapi import FastAPI
import asyncio
import uvicorn
import sqlite3
```

```python
# Database initalization

conn = sqlite3.connect("todo_db.db")
# conn = sqlite3.connect(":memory:")

def convert_to_json(data):
  return [dict(res) for res in data]

conn.row_factory = sqlite3.Row
cur = conn.cursor()
cur.execute(
  """
  CREATE TABLE todo (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL
  )
  """
)

cur.execute(
  """
  INSERT INTO todo (id, name) VALUES (null, "Poorin Note")
  """
)

cur.executemany(
  """
  INSERT INTO todo (name) VALUES (?)
  """,
  [
    ("Mulika Note",),
    ("Pongjit Note",)
  ]
)


cur.execute(
  """
  SELECT * FROM todo
  """
)

todos = cur.fetchall()

print(todos)

conn.commit() # Commit ผลการ Execute
conn.close()
```

```python
for row in todos:
  print(row["id"], row["name"], " = ", row[0], row[1])
```

```python
print(
  convert_to_json(todos)
)
```

```python
app = FastAPI()

# Route (Maybe work with controllers)
@app.get("/")
def root():
  return {
    "message": "App is running!"
  }

@app.get("/courses/{name}")
def get_course_name(name):
  return {
    "message": f"Course:  {name}"
  }
```

```python
if __name__ == "__main__":
  config = uvicorn.Config(app)
  server = uvicorn.Server(config)
  await server.serve()
```
