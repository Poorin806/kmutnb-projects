```python
# Get string input and count as following:
# - number (include dot '.')
# - Uppercase character
# - Lowercase character
# - Space
# and then display the output for each type

# Send file as .ipynb
```

```python
total_num = 0
total_upper = 0
total_lower = 0
total_space = 0
```

```python
text = input("Enter your input: ")
print(f"Your input: \"{text}\"")
```

```python
for i in range(len(text)):

  current_char = text[i]

  if (current_char.isnumeric() or current_char == '.'):
    total_num += 1
  elif (current_char.isupper()):
    total_upper += 1
  elif (current_char.islower()):
    total_lower += 1
  elif (current_char.isspace()):
    total_space += 1
```

```python
print("[Result]")
print(f"Your input: \"{text}\"")
print(f"Total numberic(s): {total_num}")
print(f"Total uppercase(s): {total_upper}")
print(f"Total lowercase(s): {total_lower}")
print(f"Total space(s): {total_space}")
```
