```python
from fastapi import FastAPI
import asyncio
import uvicorn
import sqlite3
```

```python
# Database initalization

conn = sqlite3.connect("user_db.db")
# conn = sqlite3.connect(":memory:")

def convert_to_json(data):
  return [dict(res) for res in data]

conn.row_factory = sqlite3.Row
cur = conn.cursor()
cur.execute(
  """
  CREATE TABLE user (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email VARCHAR(100) UNIQUE NOT NULL
  )
  """
)

cur.execute(
  """
  INSERT INTO user (id, name, email) VALUES (null, "Poorin Note", "poorin@example.com")
  """
)

cur.executemany(
  """
  INSERT INTO user (name, email) VALUES (?, ?)
  """,
  [
    ("Mulika Note", "mulika@example.com"),
    ("Pongjit Note", "pongjit@example.com")
  ]
)


cur.execute(
  """
  SELECT * FROM user
  """
)

users = cur.fetchall()
print(users)

conn.commit() # Commit ผลการ Execute
conn.close()
```

```python
for row in users:
  print(row["id"], row["name"], " = ", row[0], row[1])
```

```python
print(
  convert_to_json(users)
)
```

```python
app = FastAPI()

# Route (Maybe work with controllers)
@app.get("/")
def root():
  return {
    "message": "App is running!"
  }

@app.get("/courses/{name}")
def get_course_name(name):
  return {
    "message": f"Course:  {name}"
  }
```

```python
if __name__ == "__main__":
  config = uvicorn.Config(app)
  server = uvicorn.Server(config)
  await server.serve()
```
