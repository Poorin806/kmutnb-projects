```python
import sys
from PyQt6 import QtCore, QtGui, QtWidgets, uic


class MainWindow(QtWidgets.QMainWindow):
    def __init__(self):
        super().__init__()
        uic.loadUi("gui.ui", self)

        self.listView = self.findChild(QtWidgets.QListView, "listView")
        self.input = self.findChild(QtWidgets.QLineEdit,"mainInput")

        self.resultLabel = self.findChild(QtWidgets.QLabel, "resultLabel")
        self.clearButton = self.findChild(QtWidgets.QPushButton, "clearButton")
        self.addListButton = self.findChild(QtWidgets.QPushButton, "addListButton")
        self.showListButton = self.findChild(QtWidgets.QPushButton, "showListButton")
        self.removeListButton = self.findChild(QtWidgets.QPushButton, "removeListButton")

        # -------- ListView --------
        self.listData = []
        self.model = QtCore.QStringListModel(self.listData)
        self.listView.setModel(self.model)

        # -------- Button events --------
        self.listView.clicked.connect(self.handleListViewClicking)
        self.addListButton.clicked.connect(self.addItem)
        self.showListButton.clicked.connect(self.showItems)
        self.removeListButton.clicked.connect(self.removeItem)
        self.clearButton.clicked.connect(self.clearInput)

        # Init apps
        initialMessage = QtWidgets.QMessageBox(self)

        initialDialog =  initialMessage
        initialDialog.setWindowTitle("Initial Dialog")
        initialDialog.setIcon(initialMessage.Icon.Information)
        initialDialog.setText("6706021411202 - นายภูรินทร์ ทัศคร")
        initialDialog.setStandardButtons(
            initialMessage.StandardButton.Ok
        )

        initialDialog.exec()

    # -------- ListView handlers --------

    def clearInput(self):
        self.input.clear()

    def handleListViewClicking(self, index):
        item = self.model.data(index)
        self.input.setText(item)

    def addItem(self):
        text = self.input.text()
        if text:
            self.listData.append(text)
            self.model.setStringList(self.listData)
            self.input.clear()

    def showItems(self):
        print("List items:")
        for item in self.listData:
            print(item)
            self.resultLabel.setText(f"Result: {item}")

    def removeItem(self):
        index = self.listView.currentIndex()
        if index.isValid():
            self.listData.pop(index.row())
            self.model.setStringList(self.listData)

app = QtCore.QCoreApplication.instance()
if app is None : app = QtWidgets.QApplication([])
window = MainWindow()
window.show()
app.exec()
```
