```python
import sys
from PyQt6 import QtCore, QtGui, QtWidgets, uic
```

```python
class MainWindow(QtWidgets.QMainWindow):
  def __init__(self):
    super().__init__()
    uic.loadUi("main_ui.ui", self)

    # App settings
    self.setWindowFlag(
      QtCore.Qt.WindowType.WindowCloseButtonHint
    )
    self.setWindowIcon(QtGui.QIcon("logo.png"))

    # App variables
    self.userList = {}

    # Define UI Object
    self.listWidget = self.findChild(QtWidgets.QListWidget, "listWidget")

    self.userIdLineEdit = self.findChild(QtWidgets.QLineEdit, "userIdLineEdit")
    self.emailLineEdit = self.findChild(QtWidgets.QLineEdit, "emailLineEdit")
    self.fullnameLineEdit = self.findChild(QtWidgets.QLineEdit, "fullnameLineEdit")
    self.positionLineEdit = self.findChild(QtWidgets.QLineEdit, "positionLineEdit")

    self.addButton = self.findChild(QtWidgets.QPushButton, "addButton")
    self.updateButton = self.findChild(QtWidgets.QPushButton, "updateButton")
    self.removeButton = self.findChild(QtWidgets.QPushButton, "removeButton")

    # Initial values
    self.userIdLineEdit.setText("001")
    self.emailLineEdit.setText("poorin806@gmail.com")
    self.fullnameLineEdit.setText("Poorin Thatsakhon")
    self.positionLineEdit.setText("CEO")

    # Button events
    self.addButton.clicked.connect(self.addNewUser)
    self.updateButton.clicked.connect(self.editUser)
    self.removeButton.clicked.connect(self.deleteUser)

  # ---------------- End of init function ----------------


  # ---------------- Functions ----------------
  def addNewUser(self):
    userId = self.userIdLineEdit.text()
    email = self.emailLineEdit.text()
    fullname = self.fullnameLineEdit.text()
    position = self.positionLineEdit.text()

    self.userList[userId] = {
      "fullname": fullname,
      "email": email,
      "position": position
    }

    msg = f"{userId}: {fullname} ({email}) ({position})"
    newItem = QtWidgets.QListWidgetItem(msg)
    newItem.setData(
      QtCore.Qt.ItemDataRole.UserRole, userId
    )
    self.listWidget.addItem(newItem)

    self.clearInput()
  # ---------------- End of addNewUser() ----------------
  
  def editUser(self):
    userId = self.userIdLineEdit.text()
    email = self.emailLineEdit.text()
    fullname = self.fullnameLineEdit.text()
    position = self.positionLineEdit.text()

    # ถ้าไม่ได้เลือก user ที่จะแก้ไข
    currentItem = self.listWidget.currentItem()
    if not currentItem:
      QtWidgets.QMessageBox.warning(self, "Warning!", "No user selected!")
      return

    # ตรวจสอบว่าได้ มีการใส่ input ไหม
    if not userId or not email or not fullname or not position:
      QtWidgets.QMessageBox.warning(self, "Warning!", "Fill all the inputs!")
      return

    # แก้ไขที่ list widget
    msg = f"{userId}: {fullname} ({email}) ({position})"
    currentItem.setText(msg)

    # แก้ไขที่ตัวแปร userList
    self.userList[userId]["fullname"] = fullname
    self.userList[userId]["email"] = email
    self.userList[userId]["position"] = position
  # ---------------- End of editUser() ----------------

  def deleteUser(self):
    # ถ้าไม่ได้เลือก user ที่จะแก้ลบ
    currentItem = self.listWidget.currentItem()
    if not currentItem:
      QtWidgets.QMessageBox.warning(self, "Warning!", "No user selected!")
      return
    
    # ลบออกจาก Widget
    deletedId = currentItem.data(
      QtCore.Qt.ItemDataRole.UserRole
    )
    deletedIndex = self.listWidget.row(currentItem)
    self.listWidget.takeItem(deletedIndex)

    # ลบออกจาก userList
    self.userList[deletedId] = None
  # ---------------- End of deleteUser() ----------------

  def clearInput(self):
    self.userIdLineEdit.setText("")
    self.fullnameLineEdit.setText("")
    self.emailLineEdit.setText("")
    self.positionLineEdit.setText("")
  # ---------------- End of clearInput() ----------------




# Run the app
app = QtCore.QCoreApplication.instance()
if app is None: app = QtWidgets.QApplication([])
window = MainWindow()
window.show()
app.exec()
```
