```python
# ตรวจสอบว่าจำนวนจริง a มีค่าใกล้กับ b หรือไม่ โดยตรวจสอบว่า a กับ b ต่างกัน เชิงสัมพันธ์ไม่เกิน 10^(-10) หรือไม่
print("A is close to B" if abs(a - b) / max(abs(a), abs(b)) <= 1e-10 else "A is not close to B")
```

```python
def max_xy(x, y):
  return x if x > y else y

def max4(a, b, c, d):
  return max_xy(a, max_xy(b, max_xy(c, d)))
```

```python
# mx เก็บค่ามากสุดของ a, b, c, และ d เขียน Code โดยใช้ function max_xy
mx = max4(a, b, c, d)
print(f"The maximum number is {mx} ({ "A" if mx == a else "B" if mx == b else "C" if mx == c else "D" })")
```
