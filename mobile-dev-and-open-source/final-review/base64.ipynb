```python
import base64

text = "PoOrIn Th."
data = text.encode()

hex_repr = " ".join(f"{b:02X}" for b in data)
bits_8 = " ".join(f"{b:08b}" for b in data)

full_bits = "".join(f"{b:08b}" for b in data)
pad_bits = (6 - len(full_bits) % 6) % 6
bits_6_groups = [ (full_bits + "0" * pad_bits)[i:i+6] for i in range(0, len(full_bits) + pad_bits, 6) ]
bits_6 = " ".join(bits_6_groups)
dec_6 = " ".join(str(int(g, 2)) for g in bits_6_groups)

encoded = base64.b64encode(data).decode()

print("Hex representation of text:", hex_repr)
print("Bit representation of text grouped by bytes (8 bits):", bits_8)
print("Bit representation of text in a group of 6 bits:", bits_6)
print("Decimal representation of text in a group of 6 bits:", dec_6)
print("Base64 Encoded:", encoded)
```
