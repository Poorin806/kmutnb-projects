```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ===============================
# Example data (มี outlier)
# ===============================
# ค่า 120 คือ outlier
data = [12, 15, 12, 17, 18, 21, 15, 120]

# แปลงข้อมูลเป็น DataFrame
df = pd.DataFrame(data, columns=['Values'])


# ===============================
# Handle Missing Values
# ===============================
# แทน missing values ด้วย median (กรณีข้อมูลมี outlier)
df['Values'] = df['Values'].fillna(df['Values'].median())


# ===============================
# Detect Outliers using IQR Method
# ===============================
# คำนวณ Quartiles
Q1 = df['Values'].quantile(0.25)   # 25th percentile
Q3 = df['Values'].quantile(0.75)   # 75th percentile
IQR = Q3 - Q1                      # Interquartile Range

# กำหนดขอบเขต outlier
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f"Q1={Q1}, Q3={Q3}, IQR={IQR}")
print(f"Lower={lower_bound}, Upper={upper_bound}")

# ===============================
# ตรวจจับ Outliers
# ===============================
outliers = df[(df['Values'] < lower_bound) | (df['Values'] > upper_bound)]
print("\nDetected Outliers:")
print(outliers)

# ===============================
# วิธีที่ 1: Remove Outliers (ลบออก)
# ===============================
df_removed = df[(df['Values'] >= lower_bound) & (df['Values'] <= upper_bound)]
print("\nAfter Removing Outliers:")
print(df_removed)

# ===============================
# วิธีที่ 2: Replace with Median
# ===============================
median_value = df['Values'].median()
df_replaced = df.copy()
# แทนค่า outlier ด้วย median
df_replaced.loc[
    (df_replaced['Values'] < lower_bound) | (df_replaced['Values'] > upper_bound),
    'Values'
] = median_value
print("\nAfter Replacing Outliers with Median:")
print(df_replaced)


# ===============================
# วิธีที่ 3: Capping / Clipping
# ===============================
# ตัดค่าไม่ให้เกินขอบเขต IQR
df_capped = df.copy()
df_capped['Values'] = df_capped['Values'].clip(lower=lower_bound, upper=upper_bound)
print("\nAfter Capping Outliers:")
print(df_capped)


# ===============================
# Visualization: Boxplot
# ===============================
# แสดง boxplot เพื่อดู outlier ด้วยสายตา
plt.figure(figsize=(4, 3))
plt.boxplot(df['Values'], vert=True)
plt.title('Boxplot Showing Outliers (IQR Method)')
plt.ylabel('Values')
plt.grid(True)
plt.show()
```
