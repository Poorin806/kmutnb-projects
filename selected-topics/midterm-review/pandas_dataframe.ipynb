```python
# Pandas ใช้สำหรับจัดการข้อมูลในรูปแบบตาราง (DataFrame)
import pandas as pd

# key = ชื่อคอลัมน์, value = ข้อมูลในแต่ละคอลัมน์
data = {
    "name": ["Alice", "Bob", "John"],     # ข้อมูลเชิงกลุ่ม (Categorical)
    "age": [20, 22, 21],                  # ข้อมูลเชิงตัวเลข (Numerical)
    "score": [85, 90, 78]                 # คะแนน
}

# DataFrame คือโครงสร้างข้อมูลแบบตาราง (rows × columns)
df = pd.DataFrame(data)

# แสดงข้อมูลทั้งหมดใน DataFrame
print(df)
# แสดง 5 แถวแรกของข้อมูล (default = 5)
df.head()
# แสดงโครงสร้างข้อมูล: จำนวนแถว, ชนิดข้อมูล, non-null count
df.info()
# แสดงค่าสถิติของคอลัมน์เชิงตัวเลข (mean, std, min, max, quartiles)
df.describe()

# เลือกข้อมูลจากคอลัมน์เดียว → ได้ผลลัพธ์เป็น Series
df['name']
# เลือกหลายคอลัมน์พร้อมกัน → ต้องใช้ list → ผลลัพธ์เป็น DataFrame
df[['name', 'score']]
# กรองแถวข้อมูลตามเงื่อนไข (Boolean Indexing)
# เลือกเฉพาะแถวที่ age > 21
df[df['age'] > 21]


# ===============================
# Google Drive (Optional)
# ใช้เฉพาะกรณีรันบน Google Colab
# ===============================
# เชื่อมต่อ Google Drive เพื่ออ่านไฟล์จาก Drive
from google.colab import drive
drive.mount('/content/drive')
# อ่านไฟล์ CSV จาก Google Drive
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Selected Topics/iris.csv")
# แสดงข้อมูลที่อ่านจากไฟล์
df

# ===============================
# Optional / Additional Pandas commands
# ===============================
# Export selected columns to a CSV file
# ['student_id', 'total_score', 'grade'] → เลือกเฉพาะคอลัมน์ที่ต้องการ
# f"{url}/new-data-2.csv" → path สำหรับบันทึกไฟล์
df[['student_id', 'total_score', 'grade']].to_csv(f"{url}/new-data-2.csv")


# สร้างคอลัมน์ใหม่ 'pass' → True/False
# เงื่อนไข: total_score >= 90 → ผ่าน
df['pass'] = df['total_score'] >= 90
# ดูจำนวนแถวและคอลัมน์ของ DataFrame
df.shape
# ดูจำนวนแถวของคอลัมน์ grade
df['grade'].shape
# แสดงค่าที่ไม่ซ้ำในคอลัมน์ grade
df['grade'].unique()
# นับจำนวน missing values ในแต่ละคอลัมน์
df.isnull().sum()

# -------- Option 1: Fill missing values with zero --------
# ใช้เมื่อค่า 0 มีความหมายทางธุรกิจ
filled_zero = df.fillna(0)
filled_zero

# -------- Option 2: Fill missing values with mean --------
# ใช้เฉพาะคอลัมน์ attendance_percentage
# เหมาะกับข้อมูล numerical ที่ไม่ sensitive ต่อ outlier
filled_mean = df.fillna(df[['attendance_percentage']].mean())
filled_mean

# -------- Option 3: Drop rows with missing values --------
# ลบแถวที่มี missing values ทั้งหมด
# ใช้เมื่อข้อมูล missing น้อย และไม่อยากแทนค่า
dropped_row = df.dropna()
dropped_row

# -------- Option 4: Fill missing values using linear interpolation --------
# ใช้ประมาณค่าที่หายไปจากค่าก่อนหน้าและถัดไป
# เหมาะกับข้อมูลเชิงลำดับ / Time Series
filled_linear = df.interpolate(method='linear')
filled_linear
```
