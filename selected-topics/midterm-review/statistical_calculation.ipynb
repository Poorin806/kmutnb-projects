```python
# ===============================
# Import libraries สำหรับงานคำนวณทางสถิติ
# ===============================
# NumPy ใช้สำหรับการคำนวณเชิงตัวเลขและสถิติพื้นฐาน
import numpy as np
# SciPy.stats ใช้สำหรับฟังก์ชันสถิติขั้นสูง เช่น mode
import scipy.stats as stats


# ===============================
# เตรียมชุดข้อมูลตัวอย่าง (Sample Data)
# ===============================
# ข้อมูลเชิงตัวเลข (Numerical Data) ใช้แทนข้อมูลจริง เช่น คะแนนหรือปริมาณ
data = [2, 3, 5, 4, 8, 2, 5, 3, 10, 11, 2]


# ===============================
# ค่าสถิติพื้นฐาน (Descriptive Statistics)
# ===============================
# ค่าต่ำสุด (Minimum) ของข้อมูล
minimum = np.min(data)
# ค่าสูงสุด (Maximum) ของข้อมูล
maximum = np.max(data)
# ค่าเฉลี่ย (Mean) ใช้วัดค่ากลางของข้อมูล
mean = np.mean(data)
# ค่ามัธยฐาน (Median) ค่ากลางที่ไม่ถูกกระทบจาก outlier
median = np.median(data)


# ===============================
# ฐานนิยม (Mode)
# ===============================
try:
    # คำนวณ Mode โดยใช้ SciPy (รองรับข้อมูลที่มี mode มากกว่า 1 ค่า)
    mode_result = stats.mode(data, keepdims=False)
    # ถ้า mode เป็น array ให้เลือกค่าแรกมาใช้งาน
    mode = mode_result.mode[0] if isinstance(mode_result.mode, np.ndarray) else mode_result.mode
except TypeError:
    # รองรับ SciPy เวอร์ชันเก่า ที่ต้องเข้าถึงค่า mode ด้วย index
    mode = stats.mode(data)[0][0]


# ===============================
# การวัดการกระจายของข้อมูล
# ===============================
# พิสัย (Range) = ค่าสูงสุด - ค่าต่ำสุด
data_range = maximum - minimum
# ความแปรปรวน (Variance)
# ddof=1 หมายถึงการคำนวณแบบ Sample Variance (หารด้วย n-1)
variance = np.var(data, ddof=1)
# ส่วนเบี่ยงเบนมาตรฐาน (Standard Deviation)
# เป็นรากที่สองของ Variance และมีหน่วยเดียวกับข้อมูล
std_dev = np.std(data, ddof=1)


# ===============================
# Interquartile Range (IQR)
# ===============================
# Quartile ที่ 1 (Q1) คือ percentile ที่ 25
Q1 = np.percentile(data, 25)
# Quartile ที่ 3 (Q3) คือ percentile ที่ 75
Q3 = np.percentile(data, 75)
# IQR = ช่วงของข้อมูลตรงกลาง 50% (ใช้ตรวจจับ outlier)
iqr = Q3 - Q1
```
