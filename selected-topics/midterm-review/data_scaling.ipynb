```python
# ===============================
# Import libraries สำหรับ Data Cleaning และ Data Scaling
# ===============================
# NumPy ใช้สำหรับการคำนวณเชิงตัวเลข และจัดการค่า NaN
import numpy as np
# Pandas ใช้สำหรับจัดการข้อมูลในรูปแบบตาราง (DataFrame)
import pandas as pd

# ===============================
# ตัวอย่างข้อมูลที่มี Missing Values และ Outliers
# - Missing values (None, np.nan)
# - Outlier (ค่า 120 ที่สูงผิดปกติ)
data_outliers = [12, 15, 12, None, 17, 18, 21, 15, np.nan, 120]  # 120 คือ outlier


# สร้าง DataFrame เพื่อให้ง่ายต่อการทำ Data Cleaning
df_outliers = pd.DataFrame(data_outliers, columns=['Values'])


# ===============================
# การจัดการ Missing Values
# ===============================
# แทนค่าที่หายไป (NaN) ด้วย median
# ใช้ median เพราะไม่ sensitive ต่อ outlier
df_outliers['Values'] = df_outliers['Values'].fillna(
    df_outliers['Values'].median()
)


# ===============================
# การตรวจจับ Outliers ด้วย IQR Method
# ===============================
# คำนวณ Quartile ที่ 1 (25th percentile)
Q1 = df_outliers['Values'].quantile(0.25)
# คำนวณ Quartile ที่ 3 (75th percentile)
Q3 = df_outliers['Values'].quantile(0.75)
# คำนวณ Interquartile Range (IQR)
# IQR = ช่วงของข้อมูลตรงกลาง 50%
IQR = Q3 - Q1


# ===============================
# กำหนดขอบเขตของ Outliers
# ===============================
# ค่าต่ำสุดที่ยอมรับได้ (Lower bound)
lower_bound = Q1 - 1.5 * IQR
# ค่าสูงสุดที่ยอมรับได้ (Upper bound)
upper_bound = Q3 + 1.5 * IQR


# ===============================
# แทนค่า Outliers ด้วย NaN
# ===============================
# ถ้าค่าอยู่นอกช่วง [lower_bound, upper_bound]
# จะถูกแทนด้วย NaN
df_outliers['Values'] = df_outliers['Values'].where(
    (df_outliers['Values'] >= lower_bound) &
    (df_outliers['Values'] <= upper_bound),
    np.nan
)


# ===============================
# จัดการ Outliers (แทนด้วย Median)
# ===============================
# แทนค่า NaN (ที่มาจาก outlier) ด้วย median
# ทำให้ข้อมูลสะอาดและพร้อมใช้งาน
df_outliers['Values'] = df_outliers['Values'].fillna(
    df_outliers['Values'].median()
)


# ===============================
# แสดงผลลัพธ์หลังจาก Data Cleaning
# ===============================
print("Cleaned DataFrame (Outlier Handling):")
print(df_outliers)
print("-" * 50)


# ===============================
# ตัวอย่างข้อมูลสำหรับ Data Scaling
# ===============================
# ข้อมูลคะแนนนักเรียน (Numerical Data)
data_scores = [22, 15, 45, 56, 56, 65, 34, 55, 62, 84, 71, 92, 48, 98]

# แปลงข้อมูลเป็น DataFrame
df_scores = pd.DataFrame(data_scores, columns=['Values'])


# ===============================
# Normalization (Min-Max Scaling)
# ===============================
# ค่าต่ำสุดของข้อมูล
min_value = df_scores['Values'].min()
# ค่าสูงสุดของข้อมูล
max_value = df_scores['Values'].max()

# สูตร Min-Max Scaling:
# (x - min) / (max - min)
# ทำให้ข้อมูลอยู่ในช่วง 0 ถึง 1
df_scores['Normalized'] = (
    (df_scores['Values'] - min_value) /
    (max_value - min_value)
)


# ===============================
# Standardization (Z-score Scaling)
# ===============================
# ค่าเฉลี่ยของข้อมูล
mean = df_scores['Values'].mean()
# ส่วนเบี่ยงเบนมาตรฐานของข้อมูล
std_dev = df_scores['Values'].std()

# สูตร Z-score:
# (x - mean) / standard deviation
# ทำให้ข้อมูลมี mean = 0 และ std = 1
df_scores['Standardized'] = (
    (df_scores['Values'] - mean) / std_dev
)


# ===============================
# แสดงผลลัพธ์หลัง Data Scaling
# ===============================
print("\nOriginal, Normalized, and Standardized Data (Student Scores):")
print(df_scores)
```
