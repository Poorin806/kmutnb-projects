```python
# NumPy ใช้สำหรับจัดการค่าตัวเลข และค่า missing เช่น np.nan
import numpy as np
# Pandas ใช้สำหรับจัดการข้อมูลในรูปแบบตาราง (DataFrame)
import pandas as pd


# ===============================
# ตัวอย่างข้อมูลที่มี Missing Values
# ===============================
# ข้อมูลตัวเลขที่มีค่าหาย (None และ np.nan)
# ใช้จำลองข้อมูลจริงที่มักไม่สมบูรณ์
data = [12, 15, 12, None, 17, 18, 21, np.nan, 12]

# สร้าง DataFrame เพื่อให้สามารถตรวจสอบและจัดการ missing values ได้ง่าย
df = pd.DataFrame(data, columns=['Values'])


# แสดงข้อมูลทั้งหมดใน DataFrame
print("Original DataFrame:")
print(df)
# แสดงโครงสร้างข้อมูล เช่น จำนวนแถว ชนิดข้อมูล และจำนวน non-null
print("\nDataFrame Info:")
df.info()
# ตรวจสอบว่าแต่ละตำแหน่งเป็น missing value หรือไม่ (True / False)
print("\nMissing Values (isnull):")
print(df.isnull())
# นับจำนวน missing values ในแต่ละคอลัมน์
print("\nTotal Missing Values:")
print(df.isnull().sum())


# ===============================
# การจัดการ Missing Values
# ===============================

# -------- Option 1 --------
# แทนค่า missing values ด้วยค่าคงที่ (เช่น 0)
# ใช้เมื่อค่า 0 มีความหมายทางธุรกิจ
filled_zero = df.fillna(0)
print("\nData with Missing Values Filled (Zero):")
print(filled_zero)

# -------- Option 2 --------
# แทนค่า missing values ด้วยค่าเฉลี่ย (Mean)
# เหมาะกับข้อมูลที่กระจายค่อนข้างสมมาตร และไม่มี outlier มาก
filled_mean = df.fillna(df['Values'].mean())
print("\nData with Missing Values Filled (Mean):")
print(filled_mean)

# -------- Option 3 --------
# ลบแถวที่มี missing values ออกทั้งหมด
# ใช้เมื่อข้อมูลมี missing น้อย และไม่อยากบิดเบือนค่า
dropped_na = df.dropna()
print("\nData with Missing Values Dropped:")
print(dropped_na)


# -------- Option 4 --------
# ใช้การประมาณค่าแบบ Linear Interpolation
# ค่าที่หายจะถูกคำนวณจากค่าก่อนหน้าและถัดไป
# เหมาะกับข้อมูลที่มีลำดับ เช่น เวลา (time series)
interpolated = df.interpolate(method='linear')
print("\nData after Linear Interpolation:")
print(interpolated)
```
