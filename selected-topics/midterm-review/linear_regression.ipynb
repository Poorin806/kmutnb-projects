```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

# สร้าง datasets ตัวอย่าง
np.random.seed(42)
# ขนาดบ้าน (500 - 3500 sqft)
house_size = np.random.rand(100, 1) * 3000 + 500  
# ราคา = 50,000 + 150 * ขนาด + noise
house_price = 50000 + 150 * house_size + np.random.randn(100, 1) * 20000  
# แปลงเป็น DataFrame
df = pd.DataFrame(
    np.hstack((house_size, house_price)), 
    columns=['ขนาดบ้าน (sqft)', 'ราคา (USD)']
)
print("ตัวอย่างข้อมูล:")
print(df.head())

# 80% train, 20% test
X_train, X_test, y_train, y_test = train_test_split(
    house_size, house_price, test_size=0.2, random_state=42
)


# -------------------------------
# 3️⃣ สร้างและเทรนโมเดล Linear Regression
# -------------------------------
model = LinearRegression()
model.fit(X_train, y_train)  # Fit line: y = coef * X + intercept

# -------------------------------
# 4️⃣ ทำนายราคาบ้านจาก Test set
# -------------------------------
y_pred = model.predict(X_test)

# -------------------------------
# 5️⃣ ประเมินโมเดล
# -------------------------------
mse = mean_squared_error(y_test, y_pred)          # Mean Squared Error
rmse = np.sqrt(mse)                               # Root Mean Squared Error
mae = mean_absolute_error(y_test, y_pred)        # Mean Absolute Error
r2 = r2_score(y_test, y_pred)                    # R² Score
print(f"MSE (Mean Squared Error): {mse:.4f}")
print(f"RMSE (Root Mean Squared Error): {rmse:.4f}")
print(f"MAE (Mean Absolute Error): {mae:.4f}")
print(f"R² (R-squared Score): {r2:.4f}")


# -------------------------------
# 6️⃣ วาด Regression Line
# -------------------------------
plt.scatter(X_test, y_test, color='red', alpha=0.5, label='ราคาจริง')
plt.plot(X_test, y_pred, color='blue', linewidth=2, label='ราคาพยากรณ์')  # เส้น regression
plt.xlabel("ขนาดบ้าน (sqft)")
plt.ylabel("ราคา (USD)")
plt.title("พยากรณ์ราคาบ้านด้วย Linear Regression")
plt.legend()
plt.show()

# -------------------------------
# 7️⃣ ค่าพารามิเตอร์ของโมเดล
# -------------------------------
print(f"Intercept (จุดตัดแกน y): {model.intercept_[0]:.4f}")    # จุดตัดแกน y
print(f"Coefficient (ความชัน): {model.coef_[0][0]:.4f}")          # ความชัน: ราคาเปลี่ยนเท่าไรต่อ sqft


# -------------------------------
# 8️⃣ พยากรณ์ราคาบ้านตัวอย่าง
# -------------------------------
example_house_size = np.array([[2000]])  # บ้านขนาด 2000 sqft
predicted_price = model.predict(example_house_size)
print(f"ราคาพยากรณ์สำหรับบ้าน 2000 sqft: ${predicted_price[0][0]:,.2f}")
```
