```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import metrics
```

```python
label = np.random.randint(0, 4, size=100)
predict = np.random.randint(0, 4, size=100)

print(label)
print(predict)
```

```python
confusion_matrix = metrics.confusion_matrix(label, predict)
display_matrix = metrics.ConfusionMatrixDisplay(confusion_matrix, display_labels=["Cat", "Dog", "Person", "Rat"])
```

```python
display_matrix.plot()
plt.show()
```
