```python
import tkinter as tk
from tkinter import ttk
from tkinter import filedialog as fd
from PIL import Image, ImageTk

import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
# val_data
validate_data = tf.keras.preprocessing.image_dataset_from_directory(
  './data/Vegetable Images/train',
  labels='inferred',
  image_size=(180, 180),
  label_mode='categorical',
  interpolation='bilinear',
  validation_split=0.020,
  subset="validation",
  seed=1234,
  color_mode='rgb'
)

class_name = validate_data.class_names

print(class_name)
```

```python
def select_file():
  print("Select file")
  filetype=(("image file", "*.jpg"), ("All file", "*.*"))
  fn = fd.askopenfilename(filetypes=filetype, initialdir="./data/Vegetable Images/train")

  if len(fn):
    img = Image.open(fn)
    img = img.resize((180, 180))
    pic = ImageTk.PhotoImage(img)
    imgLabel.config(image=pic)
    imgLabel.image = pic

  print(fn)

  myImage = tf.keras.utils.load_img(
    fn,
    color_mode='rgb',
    target_size=(180, 180),
    interpolation='nearest',
    keep_aspect_ratio=False
  )

  input_arr = tf.keras.utils.img_to_array(myImage)
  input_arr = np.array([input_arr])

  # Predict
  predict_value = myModel.predict(input_arr)
  result = np.argmax(predict_value[0])
  percentage = predict_value[0][result] * 100

  print(f"Selected file: {fn}")
  print(f"Class: {class_name[result]}, Probability: {percentage:.2f}%")
  txtLabel.config(text=f"{class_name[result]} ({percentage:.2f}%)")
```

```python
myModel = tf.keras.models.load_model("Best.keras")

root = tk.Tk()
root.title("Demo App")
root.geometry("300x300")

img = ImageTk.PhotoImage(file="./data/Vegetable Images/train/Broccoli/0001.jpg")
imgLabel = ttk.Label(root, image=img)
imgLabel.pack()

txtLabel = ttk.Label(root, text="Predict Result")
txtLabel.pack()

openBtn = ttk.Button(root, text="Open file", command=select_file)
openBtn.pack()

root.mainloop()
```
